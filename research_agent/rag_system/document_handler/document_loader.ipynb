```python
import re
import pymupdf4llm


def pdf_to_markdown(file_path: str) -> str:
    markdown = pymupdf4llm.to_markdown(
        file_path,
        use_ocr=False,
    )

    markdown = "".join(
        c for c in markdown
        if c in "\n\t" or ord(c) >= 32
    )

    markdown = re.sub(r"\n{3,}", "\n\n", markdown)

    return markdown.strip()
```

```python
markdown = pdf_to_markdown("../Artificial_Intelligence/03_Can Edge-Deployable Vision-Language Models Identify Species_.pdf")
print(markdown)
```

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

headers_to_split_on = [
    ("#", "h1"),
    ("##", "h2"),
    ("###", "h3"),
    ("####", "h4"),
    ("#####", "h5"),
    ("######", "h6"),
]

splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on
)

sections = splitter.split_text(markdown)
```

```python
print(len(sections))
```

```python
for section in sections:
    print("content" , section.page_content)
    print("")
    print("metadata", section.metadata)
    print("\n\n\n\n")
```
